```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)
```

```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=200,
    temperature=0.3
)
```

```python
while True:
    question = input("You: ")

    if question.lower() == "exit":
        break

    prompt = f"""
You are a helpful medical AI assistant.

Question:
{question}

Answer:
"""

    output = pipe(prompt)

    print("\nAssistant:\n")
    print(output[0]["generated_text"].replace(prompt, ""))
```

```python
task2_prompt = """You are an experienced endocrinologist explaining a condition to a patient in a hospital setting.

Context: The patient has no medical background and wants a clear, accurate explanation before their appointment.

Constraints:
- Do not give specific medication names or dosages
- Do not diagnose the patient
- Keep each section short and easy to understand
- If you are not fully certain about something, say so instead of guessing

Write about diabetes using exactly this Markdown structure:

## Disease Overview
## Symptoms
## Risk Factors
## Prevention
## When to Consult a Doctor
"""

# The 'ask' function is not defined. Using the 'pipe' object for text generation.
# The 'pipe' function returns a list of dictionaries, so we need to access the generated text.
output = pipe(task2_prompt, max_new_tokens=400)
print(output[0]["generated_text"])
```

```python
task3_prompt = """Convert the clinical note into a structured summary. Follow the exact format shown in the examples.

Input:
Patient has cough and fever.
Output:
Symptoms:
- Cough
- Fever
Severity:
Moderate

Input:
Patient reports chest pain and shortness of breath.
Output:
Symptoms:
- Chest pain
- Shortness of breath
Severity:
Severe

Input:
Patient has mild headache and slight nausea.
Output:
Symptoms:
- Headache
- Nausea
Severity:
Mild

Input:
Patient has fever, cough, and oxygen saturation of 88%.
Output:
"""

output = pipe(task3_prompt, max_new_tokens=100, return_full_text=False)
print(output[0]["generated_text"])
```

```python
task4_prompt = """A patient presents with:
- Fever
- Cough
- Oxygen saturation = 88%

Think through this carefully step by step, considering severity and possible causes.
Do not show your reasoning steps. Only output the final recommendation, in one short paragraph, telling the patient what to do next.

Final recommendation:
"""

output = pipe(task4_prompt, max_new_tokens=150, return_full_text=False)
print(output[0]["generated_text"])
```

```python
task5_prompt = """You are a medical triage assistant. Read the case and respond with ONLY a valid JSON object, no explanation, no extra text, matching exactly this schema:

{
  "disease": "",
  "symptoms": [],
  "risk_level": "",
  "recommendation": ""
}

Case: Patient has fever, cough, and oxygen saturation of 88%.

JSON:
"""

output = pipe(task5_prompt, max_new_tokens=200, return_full_text=False)
raw = output[0]["generated_text"]
print("Raw model output:\n", raw)

import json as json_lib
try:
    parsed = json_lib.loads(raw)
    print("\nParsed successfully:")
    print(parsed)
except json_lib.JSONDecodeError:
    print("\nModel didn't return clean JSON - this happens often with small models like TinyLlama.")
```

| TinyLlama-1.1B-Chat-v1.0 | 1.1B | Fast prototyping, limited hardware, quick demos | Small download, low memory, fast inference | Weaker reasoning, prone to hallucination, struggles with strict formatting (as seen in Task 5) |
